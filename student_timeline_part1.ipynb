```python
import pandas as pd
import numpy as np
```

```python
studentRegistration = pd.read_csv('../../dataset/raw/studentRegistration.csv')
studentInfo = pd.read_csv('../../dataset/raw/studentInfo.csv')
studentAssessment = pd.read_csv('../../dataset/raw/studentAssessment.csv')
courses = pd.read_csv('../../dataset/raw/courses.csv')
assessments = pd.read_csv('../../dataset/raw/assessments.csv')
```

# 1) Cleaning Data

## 1.1 Student Assessment - Cleaning

```python
studentAssessment.info()
studentAssessment
```

```python
print('score unique values: ', studentAssessment['score'].unique())
```

```python
studentAssessment['score'] = pd.to_numeric(studentAssessment['score'], errors='coerce').astype('Int64')
studentAssessment = studentAssessment.dropna(subset=['score'])
print(f'score null/"?" values : {studentAssessment["score"].isnull().sum()}' )
```

```python
studentAssessment
```

## 1.2 Assessments - Cleaning

```python
assessments.info()
assessments
```

```python
print('date unique values: ', assessments['date'].unique())
```

```python
assessments['date'] = pd.to_numeric(assessments['date'], errors='coerce').astype('Int64')
assessments = assessments.dropna(subset=['date'])
print(f'date null/"?" values : {assessments["date"].isnull().sum()}' )
```

## 1.3 Student Info - Cleaning

```python
studentInfo.info()
studentInfo
```

```python
print('imd_band unique values: ', studentInfo['imd_band'].unique())
```

```python
studentInfo['imd_band'] = studentInfo['imd_band'].replace('?', np.nan)
print(f'imd null values : {studentInfo["imd_band"].isnull().sum()}')
```

Logically, the IMD value is linked with region 

```python
regions_list = studentInfo[studentInfo['imd_band'].isnull()]['region'].unique()
regions_list
```

```python
for region in regions_list:
    mode_imd_band = studentInfo[studentInfo['region'] == region]['imd_band'].mode()[0]
    studentInfo.loc[(studentInfo['imd_band'].isnull()) & (studentInfo['region'] == region), 'imd_band'] = mode_imd_band
```

```python
studentInfo.isnull().sum()
```

# 2) Merging Data

## 2.1 Merging ( Assessments - StudentAssessments )

Part-1, This has the student educational information


```python
student_data_part_1 = assessments.merge(studentAssessment, on=['id_assessment'], how='inner')
student_data_part_1.info()
student_data_part_1
```

## 2.2  Merging Student Info with Part 1

Part 2, merging Part 1 with Student Info

```python
student_data_part_2 = student_data_part_1.merge(studentInfo, on=['code_module', 'code_presentation','id_student'], how='inner')
student_data_part_2.info()
student_data_part_2
```

## 2.3 Merging Courses with Part 2 ( Checkpoint dataset )

```python
checkpoint = student_data_part_2.merge(courses, on=['code_module','code_presentation'], how='inner')
checkpoint.info()
checkpoint
```

# 3) Preparing Data

_Copying dataset_

```python
df = checkpoint.copy()
```

## 3.1 Study Status Column
[ Pass, Failed, Distinction ] have finished while [ Withdrawn ] didn't

```python
df['study_status'] = df['final_result'].apply(lambda x: 'finished' if x in ['Pass','Fail','Distinction'] else 'unfinished')
```

## 3.2 Time Step Column
Tracking student data in temporal sequnce

```python
df = df.sort_values(
    ['id_student', 'code_module', 'code_presentation', 'date']
)

group_cols = ["id_student", "code_module", "code_presentation"]

time_step_col = (
    df.groupby(
        group_cols
    ).cumcount() + 1
)

loc_date = df.columns.get_loc('date')
df.insert(loc_date, 'time_step', time_step_col)
df
```

## 3.3 Delta Days Column
Starting with 0 then calculating the difference

```python
delta_days_col = df.groupby(group_cols)['date'].diff().fillna(0)

loc_date = df.columns.get_loc('date') + 1
df.insert(loc_date,'delta_days', delta_days_col)
df
```

## 3.4 Module Progress Column
The progress should be done for the module according to the temporal sequence

```python
df['module_progress'] = (
    round(df['date'] / df['module_presentation_length'], 2)
)
df['module_progress']
```

## 3.5 Cumulative Weight Column
To solve the issue of Module's lines differences, now we can track the weight sequentially

```python
cumulative_weight_col = df.groupby(group_cols)['weight'].cumsum()

loc_weight = df.columns.get_loc('weight') + 1
df.insert(loc_weight,'cumulative_weight',cumulative_weight_col)
df
```
